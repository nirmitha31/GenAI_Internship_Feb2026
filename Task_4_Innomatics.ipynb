```python
"""
Problem 1: Mobile Recharge Validation System

This function validates a mobile recharge request based on:
1. Minimum recharge amount (₹50)
2. Allowed recharge plans
3. Retry option if invalid
"""

def validate_recharge():
    valid_plans = [199, 299, 399, 599]  # Predefined valid plans

    while True:  # Loop allows user to retry
        try:
            amount = int(input("Enter recharge amount (₹): "))

            # Rule 1: Minimum amount check
            if amount < 50:
                print(" Recharge amount must be at least ₹50.")

            # Rule 2: Valid plan check
            elif amount not in valid_plans:
                print("Invalid plan. Available plans are:", valid_plans)

            else:
                print("Recharge Successful!")
                break  # Exit loop if valid

        except ValueError:
            print("Please enter a valid numeric amount.")

# Function Call
validate_recharge()
```

```python
"""
Problem 2: Inventory Reorder Alert System

This function checks stock levels and prints:
- 'Reorder Alert' if stock < 15
- 'Stock OK' otherwise
"""

def check_inventory(products):

    # Loop through dictionary items
    for product, stock in products.items():

        if stock < 15:
            print(f"{product}: Reorder Alert (Stock = {stock})")
        else:
            print(f"{product}: Stock OK (Stock = {stock})")


# Sample Data
inventory_data = {
    "Apples": 10,
    "Bananas": 25,
    "Oranges": 5,
    "Mangoes": 18
}

# Function Call
check_inventory(inventory_data)
```

```python
"""
Problem 3: Student Result Processing System

This function:
1. Accepts list of student marks
2. Calculates average
3. Returns Pass if average ≥ 50 else Fail
"""

def process_student_result(marks):

    total = 0

    # Calculate total using loop
    for mark in marks:
        total += mark

    average = total / len(marks)

    print("Average Marks:", average)

    # Decision making
    if average >= 50:
        return "Pass"
    else:
        return "Fail"


# Sample Data
marks_list = [60, 70, 45, 80]

# Function Call
result = process_student_result(marks_list)
print("Final Result:", result)
```

```python
"""
Problem 4: Cab Fare Estimator

Fare Rules:
- Base Fare = ₹50
- ₹12 per km
- 25% extra during peak hour
- Retry option using while loop
"""

def calculate_fare(distance, peak_hour):

    base_fare = 50
    per_km_rate = 12

    fare = base_fare + (distance * per_km_rate)

    # Add peak hour charge
    if peak_hour.lower() == "yes":
        fare += fare * 0.25

    return fare


# Retry Loop
while True:
    try:
        distance = float(input("Enter distance (km): "))
        peak = input("Is it peak hour? (yes/no): ")

        total_fare = calculate_fare(distance, peak)
        print("Total Fare: ₹", total_fare)

        retry = input("Do you want to calculate again? (yes/no): ")
        if retry.lower() != "yes":
            break

    except ValueError:
        print(" Invalid input. Please enter correct values.")
```

```python
"""
Problem 5: Employee Attendance Eligibility Checker

Rules:
- Calculate attendance percentage
- Eligible if percentage ≥ 75%
"""

def check_attendance(attendance_list):

    present_days = 0

    # Count present days using loop
    for status in attendance_list:
        if status == "P":
            present_days += 1

    percentage = (present_days / len(attendance_list)) * 100

    print("Attendance Percentage:", percentage)

    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"


# Sample Data
attendance_data = ["P", "P", "A", "P", "P", "A", "P", "P"]

# Function Call
eligibility = check_attendance(attendance_data)
print("Eligibility Status:", eligibility)
```

```python
"""
Problem 6: Password Strength Checker

Rules:
- Minimum length = 8
- At least one digit
- At least one special character (@ # $)
"""

def check_password_strength(password):

    has_digit = False
    has_special = False
    special_characters = "@#$"

    # Length check
    if len(password) < 8:
        return "❌ Weak Password (Minimum 8 characters required)"

    # Loop through characters
    for char in password:
        if char.isdigit():
            has_digit = True
        if char in special_characters:
            has_special = True

    # Final validation
    if has_digit and has_special:
        return "✅ Strong Password"
    else:
        return "❌ Weak Password (Must contain digit and special character)"


# Function Call
user_password = input("Enter password: ")
print(check_password_strength(user_password))
```
