```python
#Problem 1: Smart Parking Lot Management System
def smart_parking_system(capacity, vehicle_logs):
    """
    Manages parking lot entry and exit logs.
    Calculates currently parked vehicles and checks capacity.
    """

    parked = 0
    peak = 0

    for log in vehicle_logs:
        if log == "IN":
            parked += 1
        elif log == "OUT":
            parked -= 1

        # Track peak usage
        if parked > peak:
            peak = parked

    print("Currently Parked Vehicles:", parked)
    print("Peak Parking Usage:", peak)

    if parked > capacity:
        print("Parking Status: Exceeded Capacity 🚨")
    else:
        print("Parking Status: Available")


# Sample Test
smart_parking_system(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

```python
#Problem 2: Online Food Delivery Time Estimator
def estimate_delivery_time(distance, traffic, weather):
    """
    Estimates delivery time based on distance, traffic and weather.
    """

    base_time = distance * 5   # 5 minutes per km

    # Traffic adjustment
    if traffic.lower() == "high":
        base_time += 15
    elif traffic.lower() == "medium":
        base_time += 10
    elif traffic.lower() == "low":
        base_time += 5

    # Weather adjustment
    if weather.lower() == "rainy":
        base_time += 10
    elif weather.lower() == "stormy":
        base_time += 20

    print("Estimated Delivery Time:", base_time, "minutes")


# Sample Test
estimate_delivery_time(8, "High", "Rainy")
```

```python
#Problem 3: Movie Theatre Seat Occupancy Analyzer
def seat_occupancy(total_seats, booked_seats):
    """
    Calculates occupancy percentage and show status.
    """

    booked_count = len(booked_seats)
    occupancy = (booked_count / total_seats) * 100

    print("Occupancy:", round(occupancy), "%")

    if occupancy == 100:
        print("Show Status: Housefull")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Seats Available")


# Sample Test
seat_occupancy(200, [1]*150)
```

```python
#Problem 4: Cloud Server Load Classification System
def classify_server_load(cpu_readings):
    """
    Classifies server load based on average CPU usage.
    """

    average_load = sum(cpu_readings) / len(cpu_readings)

    print("Average CPU Load:", round(average_load), "%")

    if average_load < 50:
        print("Server Status: Normal")
    elif 50 <= average_load <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")


# Sample Test
classify_server_load([45, 60, 70, 85, 90])
```

```python
#Problem 5: Smart Classroom Resource Usage Monitor
def resource_monitor(resource_usage):
    """
    Identifies overused resources in classroom.
    Overuse threshold = 8 hours
    """

    overused = []

    for resource, hours in resource_usage.items():
        if hours > 8:
            overused.append(resource)

    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")


# Sample Test
resource_monitor({
    "Projector": 6,
    "AC": 9,
    "Lights": 4
})
```

```python
#Problem 6: Online Event Registration Capacity Controller
def event_registration(capacity, registrations):
    """
    Manages event registrations and waitlist system.
    """

    if registrations <= capacity:
        print("Confirmed Registrations:", registrations)
        print("Waitlisted Users: 0")
        print("Registration Status: Open")
    else:
        waitlist = registrations - capacity
        print("Confirmed Registrations:", capacity)
        print("Waitlisted Users:", waitlist)
        print("Registration Status: Closed")


# Sample Test
event_registration(100, 105)
```
